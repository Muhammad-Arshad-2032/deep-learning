```python

# Autonomous Room Cleaner Agent
# Using Pure Python (No LangChain / LangGraph)

# 1. ROOM ENVIRONMENT
room = {
    "Bedroom": "dirty",
    "Kitchen": "dirty",
    "Living Room": "dirty",
    "Bathroom": "dirty"
}

# 2. MEMORY
memory = {
    "cleaned_rooms": [],
    "actions": []
}


# 3. PERCEPTION
def perception():
    print("\n[PERCEPTION] Observing room...")

    for area, condition in room.items():
        print(f"{area}: {condition}")

    return room


# 4. REASONING
def reasoning(observation):
    print("\n[REASONING] Deciding what to clean...")

    for area, condition in observation.items():

        if condition == "dirty":
            print(f"Decision: Clean {area}")
            return area

    print("All rooms are already clean!")
    return None


# 5. ACTION
def action(area):
    print(f"\n[ACTION] Moving to {area}...")
    print(f"Cleaning {area}...")

    room[area] = "clean"

    memory["cleaned_rooms"].append(area)
    memory["actions"].append(f"Cleaned {area}")

    print(f"{area} is now clean!")


# 6. FEEDBACK
def feedback():
    print("\n[FEEDBACK] Checking cleaning result...")

    if all(condition == "clean" for condition in room.values()):
        print("Feedback: All rooms are clean!")
        return True

    print("Feedback: Some rooms are still dirty.")
    return False


# 7. AUTONOMOUS AGENT LOOP
def room_cleaner_agent():

    print("===== ROOM CLEANER AGENT STARTED =====")

    while True:

        # Perception
        observation = perception()

        # Reasoning
        selected_area = reasoning(observation)

        # Stop if no dirty room remains
        if selected_area is None:
            break

        # Action
        action(selected_area)

        # Feedback
        is_clean = feedback()

        if is_clean:
            break

    print("\n===== CLEANING COMPLETED =====")

    print("\nMemory:")
    print("Cleaned Rooms:", memory["cleaned_rooms"])
    print("Action History:", memory["actions"])

    print("\nFinal Room Status:")
    for area, condition in room.items():
        print(f"{area}: {condition}")


# RUN AGENT
room_cleaner_agent()
```
