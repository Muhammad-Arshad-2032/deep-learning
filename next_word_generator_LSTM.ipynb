```python
input_sequnce=[]
for sentance in docs.split('\n'): ## to see the all sentance 
    tokanized_sentance=tokenizer.texts_to_sequences([sentance])[0]
    for i in range(1,len(tokanized_sentance)):
        input_sequnce.append(tokanized_sentance[:i+1])
        
```

```python
max_len=max([len(x) for x in input_sequnce])
```

```python
max_len
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
padded_input_sequance=pad_sequences(input_sequnce,maxlen=max_len,padding='pre')
```

```python
X=padded_input_sequance[:,:-1]
```

```python
from tensorflow.keras.utils import to_categorical
```

```python
Y=to_categorical(Y,num_classes=191)
Y
```

```python
Y.shape
X = X.astype('int32')
Y = Y.astype('float32')
```

```python
from tensorflow.keras.models import Sequential
from keras.layers import Embedding,LSTM,Dense
model=Sequential()
```

```python
model.add(Embedding(191,77,input_length=12))
model.add(LSTM(150))
model.add(Dense(191,activation='softmax'))
```

```python
model.compile(loss='categorical_crossentropy',optimizer='adam',metrics=['accuracy'])
```

```python
model.summary()
```

```python
model.fit(X,Y,epochs=100)
```

```python
import numpy as np
from tensorflow.keras.preprocessing.sequence import pad_sequences

def predict_next_word(sentence):
    # Sentence ko token IDs mein convert karo
    token_list = tokenizer.texts_to_sequences([sentence])[0]

    # Last 12 words/tokens rakho
    token_list = token_list[-12:]

    # Agar 12 se kam hain to padding
    token_list = pad_sequences(
        [token_list],
        maxlen=12,
        padding='pre'
    )

    # Prediction
    predicted = model.predict(token_list, verbose=0)

    # Highest probability wala word
    predicted_id = np.argmax(predicted, axis=-1)[0]

    # ID ko word mein convert karo
    for word, index in tokenizer.word_index.items():
        if index == predicted_id:
            return word

    return "Unknown"
```

```python
sentence = "artificial intelligence is"
next_word = predict_next_word(sentence)

print("Input:", sentence)
print("Predicted next word:", next_word)
```

```python
sentence = "artificial intelligence"

for i in range(10):
    next_word = predict_next_word(sentence)
    sentence += " " + next_word

print(sentence)
```
